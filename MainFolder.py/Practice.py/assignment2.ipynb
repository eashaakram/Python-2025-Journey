1. Create a dataset (10×10 with headings)

```python
import pandas as pd

data = {
    "Name": ["Ali","Sara","Ahmed","Ayesha","Bilal","Zara","Usman","Hina","Omar","Noor"],
    "Age": [20,21,19,22,20,21,23,20,22,21],
    "Gender": ["M","F","M","F","M","F","M","F","M","F"],
    "Study_Hours": [3,5,2,6,4,5,3,6,4,5],
    "Attendance": [80,90,70,95,85,88,75,92,86,89],
    "Assignment_Score": [70,85,60,90,75,88,65,92,78,84],
    "Midterm_Score": [65,80,55,88,70,85,60,90,72,83],
    "Final_Score": [75,88,60,92,78,90,68,95,80,87],
    "Sleep_Hours": [6,7,5,7,6,7,5,8,6,7],
    "Result": ["Pass","Pass","Fail","Pass","Pass","Pass","Fail","Pass","Pass","Pass"]
}

df = pd.DataFrame(data)
df.to_csv("students.csv", index=False)
```

2. Load and explore data using Pandas


```python
import pandas as pd

df = pd.read_csv("students.csv")

# Basic info
df.head()              # first 5 rows
df.tail()              # last 5 rows
df.info()              # data types
df.describe()          # statistics

# Selection
df["Study_Hours"]
df[["Study_Hours","Final_Score"]]

# Filtering
df[df["Final_Score"] > 80]

# Grouping
df.groupby("Gender")["Final_Score"].mean()

# Sorting
df.sort_values(by="Final_Score", ascending=False)
```

3. Perform numerical analysis using NumPy


```python
import numpy as np

scores = df["Final_Score"].values

np.mean(scores)     # average
np.median(scores)   # median
np.std(scores)      # standard deviation
np.min(scores)      # minimum
np.max(scores)      # maximum

# Array operations
arr = np.array(scores)
arr * 2
arr + 5
```

4. Create multiple visualizations using Matplotlib and Seaborn


```python
import matplotlib.pyplot as plt

# Line plot
plt.plot(df["Final_Score"])
plt.title("Final Scores Trend")
plt.show()

# Bar chart
plt.bar(df["Name"], df["Final_Score"])
plt.xticks(rotation=45)
plt.title("Student Scores")
plt.show()
```

```python
import seaborn as sns

# Histogram
sns.histplot(df["Final_Score"])
plt.show()

# Boxplot
sns.boxplot(x=df["Final_Score"])
plt.show()

# Heatmap
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.show()
```

5. Apply Scipy for statistical analysis.


```python
from scipy import stats

# Correlation
stats.pearsonr(df["Study_Hours"], df["Final_Score"])

# T-test example
group1 = df[df["Gender"]=="M"]["Final_Score"]
group2 = df[df["Gender"]=="F"]["Final_Score"]

stats.ttest_ind(group1, group2)
```
