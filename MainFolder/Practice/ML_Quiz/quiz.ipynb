Q6: Which publisher has the highest total units sold?

```python
df.columns = df.columns.str.strip()

publisher_units = df.groupby("Publisher")["units sold"].sum()

publisher_units.sort_values(ascending=False).head()
```

Q7: Find the correlation between gross sales and publisher revenue

```python
correlation = df["gross sales"].corr(df["publisher revenue"])

print("Correlation:", correlation)

if abs(correlation) >= 0.7:
    print("Strong correlation")
else:
    print("Weak correlation")
```
