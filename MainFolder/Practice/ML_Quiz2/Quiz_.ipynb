Dataset Load

```python
import pandas as pd

df = pd.read_csv("Books_Data_Clean.csv")

df.head()
```

Q1: Which genre appears most frequently in the dataset?

```python
df["genre"].value_counts()
```

Q2: Average (mean) Book_average_rating 

```python
df["Book_average_rating"].mean()
```

Q3: Publisher has highest TOTAL units sold?

```python
publisher_units = df.groupby("Publisher")["units sold"].sum()
publisher_units.head()
```

Q4: Missing values

```python
# df.isnull().sum()
# Book Name:
# Missing book names are replaced with "Unknown" because the book name is categorical/text data.
df["Book Name"] = df["Book Name"].fillna("Unknown")

# language_code:
# Missing language codes are replaced with "Unknown" because language code is categorical data.
df["language_code"] = df["language_code"].fillna("Unknown")

# Publishing Year:
# Missing publishing years are replaced with the median because it is numerical data
# and the median is less affected by extreme values.
df["Publishing Year"] = df["Publishing Year"].fillna(
    df["Publishing Year"].median()
)

df.head()
```

Q5: Sort dataset 

```python
# Sort the DataFrame by 'Publishing Year' from smallest to largest
df_sorted_by_year = df.sort_values(by='Publishing Year', ascending=True)
print("DataFrame sorted by 'Publishing Year' (smallest to largest):")
display(df_sorted_by_year.head(10))
```
