```python
# Source URL: https://www.kaggle.com/datasets/srisyra02/online-food-ordering-dataset
# Format: CSV, 388 rows, 14 columns

import pandas as pd
df = pd.read_csv("online food delivery dataset.csv")
print("Original Dataset Shape:", df.shape)
display(df.head())
```

```python
# 1. Check dataset size
print("Dataset Shape:", df.shape)
# 2. Check missing values
print("\nMissing Values:")
print("[ " + ", ".join(f"{col} : {val}" for col, val in df.isnull().sum().items()) + " ]")
# 3. Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Check data types
print("\nData Types:")
print(df.dtypes)
```

```python
# Check unique values in each column
print("\nUnique Values:")
for column in df.columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python

# Show only the rows that are exact duplicates
exact_duplicates = df[df.duplicated()]

print("Exact duplicate rows:", len(exact_duplicates))

display(exact_duplicates.head(10))
```

```python
# Remove exact duplicate rows
df_clean = df.drop_duplicates().copy()

print("Original rows:", len(df))
print("Rows after removing duplicates:", len(df_clean))
print("Remaining duplicate rows:", df_clean.duplicated().sum())
```

```python
df_clean = df_clean.rename(
    columns={"Unnamed: 13": "Additional_Response"}
)

print(df_clean.columns.tolist())
```

```python
print("Rows:", len(df_clean))
print("Remaining duplicates:", df_clean.duplicated().sum())
```

```python
text_columns = df_clean.select_dtypes(include="object").columns

for column in text_columns:
    print(f"\n{column}:")
    print(df_clean[column].unique())
```

```python
text_columns = df_clean.select_dtypes(include="object").columns

for column in text_columns:
    df_clean[column] = df_clean[column].str.strip()

print("Extra spaces removed from text columns.")
```

```python
df_clean["Occupation"] = df_clean["Occupation"].replace({
    "Self Employeed": "Self Employed"
})

print("Text cleaning completed.")
```

```python
checks = {
    "Age below 18": int((df_clean["Age"] < 18).sum()),
    "Invalid family size": int((df_clean["Family size"] <= 0).sum()),
    "Invalid latitude": int((~df_clean["latitude"].between(-90, 90)).sum()),
    "Invalid longitude": int((~df_clean["longitude"].between(-180, 180)).sum())
}

print("Invalid value checks:")
print(checks)
```

```python
print("Missing values before handling:")
print(df_clean.isnull().sum())

if df_clean.isnull().sum().sum() == 0:
    print("No missing values found. No imputation required.")
```

```python
df_clean["Age_Group"] = pd.cut(
    df_clean["Age"],
    bins=[17, 25, 35, 50, 100],
    labels=["18-25", "26-35", "36-50", "51+"]
)

df_clean["Is_Student"] = (
    df_clean["Occupation"].eq("Student").astype(int)
)

print("Derived features created.")
display(df_clean[["Age", "Age_Group", "Occupation", "Is_Student"]].head())
```

```python
rows_before = len(df_clean)

df_clean = df_clean[
    df_clean["Age"].between(18, 100)
    & df_clean["Family size"].gt(0)
    & df_clean["latitude"].between(-90, 90)
    & df_clean["longitude"].between(-180, 180)
].copy()

df_clean.reset_index(drop=True, inplace=True)

print("Rows before filtering:", rows_before)
print("Rows after filtering:", len(df_clean))
print("Rows removed:", rows_before - len(df_clean))
```

```python
print("Final dataset shape:", df_clean.shape)
print("Missing values:", df_clean.isnull().sum().sum())
print("Remaining duplicates:", df_clean.duplicated().sum())
print("Final data types:")
print(df_clean.dtypes)

display(df_clean.head())
```

```python
df_clean.to_csv("transformed_food_delivery.csv", index=False)

print("Transformed dataset saved successfully.")
```

```python
import sqlite3

connection = sqlite3.connect("food_delivery.db")

df_clean.to_sql(
    "food_delivery",
    connection,
    if_exists="replace",
    index=False
)

print("Data loaded successfully into SQLite.")
```

```python
loaded_df = pd.read_sql_query(
    "SELECT * FROM food_delivery",
    connection
)

print("Transformed rows:", len(df_clean))
print("Loaded rows:", len(loaded_df))

assert len(df_clean) == len(loaded_df)

from pandas.testing import assert_frame_equal

expected_df = df_clean.reset_index(drop=True).copy()
actual_df = loaded_df.reset_index(drop=True).copy()

for col in expected_df.columns:
    if str(expected_df[col].dtype) == "category":
        expected_df[col] = expected_df[col].astype(str)

assert_frame_equal(
    expected_df,
    actual_df,
    check_dtype=False
)

print("Verification successful: loaded data matches transformed data.")

connection.close()
```
